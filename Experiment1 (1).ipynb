```python
from google.colab import userdata
import requests

api_key = userdata.get("GEMINI_API_KEY")

if not api_key:
    raise ValueError("Gemini API key not found.")

print("API key loaded successfully!")
```

```python
prompt = """
Explain Artificial Intelligence in simple words.

Give five real-world examples suitable
for a first-year engineering student.
"""

print("Prompt created successfully!")
print(prompt)
```

```python
import requests

models_url = "https://generativelanguage.googleapis.com/v1beta/models"

models_response = requests.get(
    models_url,
    headers={"x-goog-api-key": api_key}
)

if models_response.status_code == 200:
    models_data = models_response.json()

    for model in models_data.get("models", []):
        methods = model.get("supportedGenerationMethods", [])

        if "generateContent" in methods:
            print(model["name"])
else:
    print(models_response.status_code)
    print(models_response.text)
```

```python
from transformers import pipeline

generator = pipeline("text-generation", model="Qwen/Qwen2.5-3B-Instruct", device_map="auto")
output = generator(prompt, max_new_tokens=512)
print(output[0]["generated_text"])
```
